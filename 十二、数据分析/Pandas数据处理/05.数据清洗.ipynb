```python
import pandas as pd

df = pd.read_csv("../data/movies_info3.csv")

#空值处理
df.isnull()  #查看是否有空值
df1  = df.dropna()  #删除空值所在行并返回删除后的df对象（源df表格数据不变）
df2 = df.dropna(axis=1)  #删除空值所在列并返回删除后的df对象（源df表格数据不变）

df3 = df.fillna('------')  #将空值替换为'------'
df4 = df.ffill()  #拿前行所在列的值填充空值
df5 = df.bfill()  #拿后行所在列的值填充空值
```

```python
# 重复值处理
s = df.duplicated()   #查看是否有相同行并返回Series对象
s1 = df.duplicated(subset=['类型'])    #查看某列是否有重复值
df.drop_duplicates(subset=['类型'], keep='first')   # 按指定列判断某行是否存在重复值，存在默认只保留首个非重复行（keep指定删除规则）
```

```python
# 异常值处理
df[df['时常'] < 1]     # 筛选出时常小于1的电影
df.drop(df[df['时常'] < 1].index)   # 删除时常小于1的电影
df['时常'] = df['时常'].abs()   # 将时常的负值转换为正值
```

```python
# 数据格式处理
df[df['上映时间'].str.match(r'^\d{4}/\d{2}/\d{2}$')]  # 筛选出上映时间格式为yyyy/mm/dd的电影
df['上映时间'] = df['上映时间'].str.replace('/', '-')  # 将上映时间格式转换为yyyy-mm-dd
df
```
