```python
import pandas as pd

df = pd.read_csv('../data/movies_info3_1.csv', usecols=['电影名', '上映时间', '评分', '语言'], index_col='电影名')

#查看数据
df.head(10)  #查看前10行
df.tail(10)  #查看后10行
df.describe()  #查看数值列的统计描述信息
df.info()  # 查看表结构
df.shape  #查看表的行列数
df.columns  # 查看列名

#选择数据
# 选择列
df['上映时间']  #查看单列
df.上映时间

df[['上映时间', '语言']]  # 查看多列

#选择行
df.iloc[0:10:1]  #基于隐藏的固定行号切片数据，不包含结束行（行号默认从0开始）
df.loc['安昂传奇：最后的气宗':'教父2':1]  #基于索引列切片行记录，包含结束索引行

#过滤数据
#单列过滤
df[df['评分'] >= 80]  #筛选出评分大于等于80的电影
df[df['语言'].isin([' 英语', ' 日语'])]  #筛选出语言为日语或英语的电影
df[df['评分'].between(80, 100, inclusive='both')]  #筛选出评分在80-100之间的电影(inclusive='both'表示包含80和100)

#多列过滤
df[(df['上映时间'] >= '2026-08-21') & (df['评分'] == 87)]  #筛选出上映时间在2026-08-21之后且评分为87的电影
df[(df['语言'].isin([' 英语', ' 日语'])) & (df['评分'] >= 88)]  #筛选出语言为日语或英语且评分大于等于88的电影
```
